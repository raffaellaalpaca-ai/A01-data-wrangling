```python
# Your Phase 1 solution to the problem goes here.
import pandas as pd
import matplotlib.pyplot as plt
airbnb = pd.read_csv('https://raw.githubusercontent.com/sds3001/A01-data-wrangling/main/data/airbnb_hw.csv')

#shows first 5 rows to see the columns
airbnb.head()
#to show price column
airbnb['Price'].head()
```

```python
#to inspect the bottom part
airbnb['Price'].tail()
```

```python
airbnb['Price'].dtype
```

```python
#looking for prices that have commas
airbnb[airbnb['Price'].str.contains(',')]
```

```python
#price values that are problematic
prices_w_commas = airbnb[airbnb['Price'].str.contains(',')]
prices_w_commas['Price'].head()
```

```python
#removimg commas
airbnb['Price'] = airbnb['Price'].str.replace(',', '')
#checking again
airbnb['Price'].head()
```

```python
airbnb['Price'].dtype
```

```python
#dtype is still an object
airbnb['Price'] = pd.to_numeric(airbnb['Price'])
airbnb['Price'].dtype
```

```python
airbnb['Price'].head()
```

Prices over 999 contained commas such as 1,000 or 1,990. This caused the price column to be stores as object instead of being a numeric type. I removed the commas and converted the column to numeric values

Question 2 - Part 2

```python
#loading dataset

police = pd.read_csv('https://raw.githubusercontent.com/sds3001/A01-data-wrangling/main/data/mn_police_use_of_force.csv')
```

```python
police.head()
```

```python
police['subject_injury'].value_counts(dropna=False)
```

```python
#before deciding what to do with missing values, im gonna check if related to force_type
pd.crosstab(police['subject_injury'], police['force_type'])
```

```python
#making a temp version for missing values
injury_check = police['subject_injury'].fillna('Missing')
```

```python
pd.crosstab(injury_check, police['force_type'])
```

```python
#calc the overall proportion of subject_injury that is missing
police['subject_injury'].isna().mean()
```

Around 76% of the subject_injury values are missing. This is concerning because most of the data is not available , so its hard to make a promising conclusion about injuries.Also the missing values appear a lot for some force tyoes, so the missing data may not be fully random.

```python
#cleaning column
police['injury_clean']= police['subject_injury']
```

```python
police['injury_clean'].value_counts(dropna=False)
```

The recorded subject_injury values were already labeled as yes and no. I decided to keeo the missing values as missing because it doesnt mean that no injuries occurred.

```python
#cross tab using the clean column
pd.crosstab(police['injury_clean'], police['force_type'])
```

The cross tab shows that injuries are common for some tyoe force than others. Bodily force has many injury cases, while on the other hand, other force types have fewer.
